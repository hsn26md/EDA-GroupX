Kolom yang saya duga MNAR: `caffeine_post_5pm_mg`. Tidak ada sel kosong di dataset ini, tetapi 58% barisnya bernilai
`0`, dan nilai terkecil di atas nol adalah 20 mg. Bila responden yang tidak mau menyebutkan konsumsi kafeinnya
dicatat sebagai `0`, maka kekosongannya bergantung pada nilai kafein yang sebenarnya, itulah ciri MNAR. Dugaan ini
tidak bisa dibuktikan dari datanya sendiri, karena `0` karena "tidak minum" dan `0` karena "tidak mengisi" tampak
persis sama di data. Persentase nol pun hampir sama di semua kelompok pekerjaan, kronotipe, dan jenis kelamin,
sehingga tidak ada kolom lain yang bisa menjelaskannya. Yang bisa memastikan hanya pembuat dataset atau formulir
pengisiannya.
