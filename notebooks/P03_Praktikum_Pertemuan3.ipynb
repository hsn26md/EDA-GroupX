```python
# TODO D1 — susun tabel penandaan Anda sendiri
# Buat dictionary bernama PENANDAAN berisi pasangan:
#     'nama_kolom': ('kategori', 'tindakan')
# Tindakan yang tersedia: 'hapus', 'hash', 'generalisasi', 'pertahankan'
# Kolom yang tidak Anda daftarkan dianggap 'bukan data pribadi' + 'pertahankan'.
#
# Lalu bangun DataFrame 'tanda' dengan kolom: nama_kolom, kategori, tindakan
# dan cetak jumlah kolom per kategori.
# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom

PENANDAAN = {
    # --- Pengenal langsung (direct identifier)
    'id_peserta':    ('pengenal langsung', 'hapus'),
    'nama_lengkap':  ('pengenal langsung', 'hapus'),
    'nik_dummy':     ('pengenal langsung', 'hapus'),
    'no_hp':         ('pengenal langsung', 'hash'),
    'email':         ('pengenal langsung', 'hash'),
    'alamat_jalan':  ('pengenal langsung', 'generalisasi'),

    # --- Kuasi-identifier
    'kelurahan':             ('kuasi-identifier', 'generalisasi'),
    'kecamatan':             ('kuasi-identifier', 'generalisasi'),
    'tanggal_lahir':         ('kuasi-identifier', 'generalisasi'),
    'jenis_kelamin':         ('kuasi-identifier', 'pertahankan'),
    'pendidikan_terakhir':   ('kuasi-identifier', 'pertahankan'),
    'nama_usaha':            ('kuasi-identifier', 'hash'),

    # --- Data spesifik/sensitif
    'status_disabilitas':          ('data spesifik', 'pertahankan'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'generalisasi'),

}

# Bangun DataFrame dari seluruh kolom df
tanda = pd.DataFrame({
    'nama_kolom': df.columns
})

def ambil_kategori(nama):
    return PENANDAAN.get(nama, ('bukan data pribadi', 'pertahankan'))[0]

def ambil_tindakan(nama):
    return PENANDAAN.get(nama, ('bukan data pribadi', 'pertahankan'))[1]

tanda['kategori'] = tanda['nama_kolom'].apply(ambil_kategori)
tanda['tindakan'] = tanda['nama_kolom'].apply(ambil_tindakan)

print(tanda['kategori'].value_counts())
tanda
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'
import hashlib

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    else:
        return hashlib.sha256(garam.encode() + str(nilai).encode()).hexdigest()[:12]

print(samarkan('test@contoh.com', GARAM))


```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
print(samarkan('test@contoh.com', GARAM))
print(samarkan('test@contoh.com', GARAM))
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
print("Nilai yang berbeda")
print(samarkan('test@contoh2.com', GARAM))
print(samarkan('test@contoh3.com', GARAM))

```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup
def cek_k(data, kolom):
    ukuran_grup = data.groupby(kolom, observed=True).size()

    k = ukuran_grup.min()
    grup_berisi_1 = (ukuran_grup == 1).sum()
    total_grup = ukuran_grup.shape[0]

    return pd.Series({
        'k': k,
        'grup_berisi_1': grup_berisi_1,
        'total_grup': total_grup
    })

hasil = cek_k(aman, ['jenis_kelamin', 'kelompok_umur', 'kecamatan', 'pendidikan_terakhir'])
print(hasil)
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

# TODO F2 — ukur risiko SEBELUM penanganan
kolom_qi = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']

hasil_sebelum = cek_k(df, kolom_qi)
print(hasil_sebelum)

# Hasil
assert hasil_sebelum['k'] == 1, f"k = {hasil_sebelum['k']}, harusnya 1"
assert hasil_sebelum['grup_berisi_1'] == 620, f"grup_berisi_1 = {hasil_sebelum['grup_berisi_1']}, harusnya 620"
print("\nVerifikasi lolos ✓ — semua 620 baris unik, k=1")
```

Meskipun dari 4 kolom itu tidak menampilkan pengenal langsung; tetapi bagi orang-orang yang kenal dekat bisa mengetahuinya itu punya siapa. Pada akhirnya, bisa menyebabkan informasi-informasi penting/sensitif lain bocor.

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1

kolom_qi_sesudah = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']

hasil_sesudah = cek_k(aman, kolom_qi_sesudah)
print(hasil_sesudah)

# Verifikasi
assert hasil_sesudah['k'] == 1, f"k = {hasil_sesudah['k']}, harusnya masih 1"
assert hasil_sesudah['grup_berisi_1'] < hasil_sebelum['grup_berisi_1'], "Jumlah grup berisi 1 tidak turun"

print("\nPerbandingan:")
print("Sebelum — grup berisi 1:", hasil_sebelum['grup_berisi_1'])
print("Sesudah — grup berisi 1:", hasil_sesudah['grup_berisi_1'])
print("\nVerifikasi lolos ✓")
```

```python
# TODO F4 — TANTANGAN: capai k >= 5

# --- Siapkan variasi kolom ---
# Kecamatan kecil (< 30 peserta) digabung menjadi 'Lainnya'
jml_kec = aman['kecamatan'].value_counts()
kec_kecil = jml_kec[jml_kec < 30].index
aman['kecamatan_v2'] = aman['kecamatan'].where(~aman['kecamatan'].isin(kec_kecil), 'Lainnya')

# Tiga versi kelompok umur: 5 kelompok, 3 kelompok, 2 kelompok
aman['umur_5'] = pd.cut(aman['usia'], bins=[0, 25, 35, 45, 55, 150],
                        labels=['<25', '25-34', '35-44', '45-54', '55+'], right=False)
aman['umur_3'] = pd.cut(aman['usia'], bins=[0, 35, 55, 150],
                        labels=['<35', '35-54', '55+'], right=False)
aman['kelompok_umur_v3'] = pd.cut(aman['usia'], bins=[0, 27, 150],
                                  labels=['<27', '27+'], right=False)

# --- Daftar percobaan: (catatan pilihan, daftar kolom QI) ---
percobaan = [
    ('Awal: kecamatan, jenis kelamin, pendidikan, umur 5 kelompok',
     ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'umur_5']),
    ('Umur dilebarkan jadi 3 kelompok',
     ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'umur_3']),
    ('Umur 3 kelompok + kecamatan kecil digabung',
     ['kecamatan_v2', 'jenis_kelamin', 'pendidikan_terakhir', 'umur_3']),
    ('Keluarkan pendidikan_terakhir',
     ['kecamatan', 'jenis_kelamin', 'umur_5']),
    ('Hanya kecamatan_v2 + umur 3 kelompok',
     ['kecamatan_v2', 'umur_3']),
    ('Hanya kecamatan_v2 + umur 2 kelompok (<27 / 27+)',
     ['kecamatan_v2', 'kelompok_umur_v3']),
]

# --- Jalankan semua dan catat ---
baris = []
for catatan, kolom in percobaan:
    h = cek_k(aman, kolom)
    baris.append({
        'pilihan'      : catatan,
        'kolom_QI'     : ', '.join(kolom),
        'k'            : int(h['k']),
        'grup_berisi_1': int(h['grup_berisi_1']),
        'total_grup'   : int(h['total_grup']),
        'k>=5'         : 'ya' if h['k'] >= 5 else 'tidak',
    })

catatan_f4 = pd.DataFrame(baris)
print(catatan_f4.to_string(index=False))

```
