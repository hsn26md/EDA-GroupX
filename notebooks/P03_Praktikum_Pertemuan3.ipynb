```python
import os, shutil

# 1. Buat struktur folder
os.makedirs('data/raw', exist_ok=True)
os.makedirs('data/processed', exist_ok=True)

# 3. Baca dari raw
df = pd.read_csv('data/raw/bedtime_screentime_sleep_debt.csv')
```

```python
# 4. Jalankan pola E1, E2, E4 (hapus, hash, generalisasi) — sesuaikan dengan PENANDAAN kolom sleep

PENANDAAN_SLEEP = {
    # --- Pengenal langsung (direct identifier) ---
    'user_id':         ('pengenal langsung', 'hash'),

    # --- Kuasi-identifier ---
    'age':             ('kuasi-identifier', 'generalisasi'),
    'gender':          ('kuasi-identifier', 'pertahankan'),
    'occupation_type': ('kuasi-identifier', 'pertahankan'),

    # --- Bukan data pribadi (perilaku, bukan identitas) ---
    'chronotype':      ('bukan data pribadi', 'pertahankan'),
}

# Bangun DataFrame dari seluruh kolom df
tanda_sleep = pd.DataFrame({
    'nama_kolom': df.columns
})

def ambil_kategori(nama):
    return PENANDAAN_SLEEP.get(nama, ('bukan data pribadi', 'pertahankan'))[0]

def ambil_tindakan(nama):
    return PENANDAAN_SLEEP.get(nama, ('bukan data pribadi', 'pertahankan'))[1]

tanda_sleep['kategori'] = tanda_sleep['nama_kolom'].apply(ambil_kategori)
tanda_sleep['tindakan'] = tanda_sleep['nama_kolom'].apply(ambil_tindakan)

print(tanda_sleep['kategori'].value_counts())
tanda_sleep
```

```python
kolom_dihapus = [kolom for kolom, (kategori, tindakan) in PENANDAAN_SLEEP.items() if tindakan == 'hapus']

# aman_sleep = df, MINUS kolom-kolom yang harus dihapus
aman_sleep = df.drop(columns=kolom_dihapus)

print("Kolom dibuang :", len(kolom_dihapus), "→", kolom_dihapus)
print("Kolom tersisa :", aman_sleep.shape[1], "dari", df.shape[1])

aman_sleep.head()
```

```python
# Hash
GARAM = 'ganti-dengan-kalimat-rahasia-tim'
import hashlib

def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan
    else:
        return hashlib.sha256(garam.encode() + str(nilai).encode()).hexdigest()[:12]

print(samarkan('USR-00001', GARAM))
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
print(samarkan('USR-00001', GARAM))
print(samarkan('USR-00001', GARAM))
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda
print("Nilai yang berbeda")
print(samarkan('USR-00003', GARAM))
print(samarkan('USR-00004', GARAM))
```

```python
# --- E4: GENERALISASI ---
batas_umur = [18, 25, 35, 45, 55, 66]
label_umur = ['18-24', '25-34', '35-44', '45-54', '55-65']
aman_sleep['kelompok_umur'] = pd.cut(df['age'], bins=batas_umur, labels=label_umur, right=False)
aman_sleep = aman_sleep.drop(columns=['age'], errors='ignore')

print("\nE4 — Distribusi kelompok umur:")
print(aman_sleep['kelompok_umur'].value_counts().sort_index())

aman_sleep.head()
```

```python
# 5. Jalankan pola F1-F5 (ukur k) — pakai fungsi cek_k yang sudah ada

def cek_k(data, kolom):
    ukuran_grup = data.groupby(kolom, observed=True).size()
    k = ukuran_grup.min()
    grup_berisi_1 = (ukuran_grup == 1).sum()
    total_grup = ukuran_grup.shape[0]
    return pd.Series({'k': k, 'grup_berisi_1': grup_berisi_1, 'total_grup': total_grup})
```

```python
# F2 — sebelum penanganan (pakai df_sleep asli)
kolom_qi_sebelum = ['age', 'gender', 'occupation_type', 'chronotype']
hasil_sebelum_sleep = cek_k(df, kolom_qi_sebelum)
print("F2 — Sebelum:")
print(hasil_sebelum_sleep)

# F3 — sesudah generalisasi
kolom_qi_sesudah = ['kelompok_umur', 'gender', 'occupation_type', 'chronotype']
hasil_sesudah_sleep = cek_k(aman_sleep, kolom_qi_sesudah)
print("\nF3 — Sesudah:")
print(hasil_sesudah_sleep)
```

```python
print("Status k saat ini:", hasil_sesudah_sleep['k'])

if hasil_sesudah_sleep['k'] < 5:
    # Cek kombinasi mana yang masih berisiko
    ukuran_grup = aman_sleep.groupby(kolom_qi_sesudah, observed=True).size().reset_index(name='jumlah')
    berisiko = ukuran_grup[ukuran_grup['jumlah'] < 5].sort_values('jumlah')
    print("Kombinasi berisiko (<5):")
    print(berisiko.head(10))

    # Coba lebarkan kelompok umur jadi 3 kategori
    batas_lebar = [9, 25, 80]
    label_lebar = ['10 - 25', '25+']
    aman_sleep['kelompok_umur_v2'] = pd.cut(df['age'], bins=batas_lebar, labels=label_lebar, right=False)

    hasil_v2 = cek_k(aman_sleep, ['kelompok_umur_v2', 'gender', 'occupation_type', 'chronotype'])
    print("\nPercobaan — umur dilebarkan jadi 3 kelompok:")
    print(hasil_v2)

    if hasil_v2['k'] < 5:
        # Coba keluarkan chronotype dari daftar quasi-identifier
        hasil_v3 = cek_k(aman_sleep, ['kelompok_umur_v2', 'gender', 'occupation_type'])
        print("\nPercobaan — tanpa chronotype:")
        print(hasil_v3)
```

```python
tabel_silang_sleep = pd.crosstab(aman_sleep['kelompok_umur'], aman_sleep['occupation_type'])
print("Tabel silang kelompok_umur x occupation_type:")
print(tabel_silang_sleep)

sel_kecil_sleep = (tabel_silang_sleep > 0) & (tabel_silang_sleep < 5)
jumlah_sel_kecil_sleep = sel_kecil_sleep.sum().sum()
print(f"\nJumlah sel terisi tapi < 5 peserta: {jumlah_sel_kecil_sleep}")
```
